Let's calculate first the difference of distances of the associated tower assets from the focus/reference tower asset.

```python
from geopy.distance import geodesic
import numpy as np

def calculate_distances_to_reference(df):
    df_copy = df.copy()
    df_copy['distance_to_reference'] = np.nan

    for group_name, group_df in df_copy.groupby('focus_asset_id'):
        reference_record = group_df[group_df['associated_asset_id'].isna()]

        if not reference_record.empty:
            ref_lat = reference_record['latitude'].iloc[0]
            ref_lon = reference_record['longitude'].iloc[0]
            reference_coords = (ref_lat, ref_lon)

            for index, row in group_df.iterrows():
                if pd.notna(row['associated_asset_id']):
                    record_coords = (row['latitude'], row['longitude'])
                    distance = geodesic(reference_coords, record_coords).meters
                    df_copy.loc[index, 'distance_to_reference'] = distance
    return df_copy

df_with_distances = calculate_distances_to_reference(df)
df_with_distances
```

Let's find the difference in AGL of the reference tower against each associated tower.

```python
def calculate_agldiff_to_reference(df):
    df_copy = df_with_distances.copy()
    df_copy['agldiff_to_reference'] = np.nan

    for group_name, group_df in df_copy.groupby('focus_asset_id'):
        reference_record = group_df[group_df['associated_asset_id'].isna()]

        if not reference_record.empty:
            ref_agl = reference_record['agl'].iloc[0]

            for index, row in group_df.iterrows():
                if pd.notna(row['associated_asset_id']):
                    record_agl = row['agl']
                    distance = ref_agl - record_agl
                    df_copy.loc[index, 'agldiff_to_reference'] = distance
    return df_copy

df_with_differences = calculate_agldiff_to_reference(df_with_distances)
df_with_differences
```

```python
def find_case_1_candidates(df_with_differences):

    case_1_records = []

    grouped = df_with_differences.groupby('focus_asset_id')

    for focus_asset_id, group in grouped:
        reference_record_df = group[group['associated_asset_id'].isnull()]
        paired_records_df = group[group['associated_asset_id'].notnull()]

        if len(reference_record_df) == 1:
            reference_record = reference_record_df.iloc[0]
            ref_fcc = reference_record['fcc_asr_number']
            ref_faa = reference_record['faa_study_number']

            if pd.notnull(ref_fcc) and pd.notnull(ref_faa):
                matching_paired_records = paired_records_df[
                    (paired_records_df['fcc_asr_number'] == ref_fcc) &
                    (paired_records_df['faa_study_number'] == ref_faa) &
                    (paired_records_df['fcc_asr_number'].notnull()) &
                    (paired_records_df['faa_study_number'].notnull())
                ]

                if not matching_paired_records.empty:
                    case_1_records.append(reference_record_df)
                    case_1_records.append(matching_paired_records)

    if case_1_records:
        case_1_candidates = pd.concat(case_1_records, ignore_index=True)
        return case_1_candidates.drop_duplicates()
    else:
        return pd.DataFrame(columns=df_with_differences.columns)

case_1_candidates = find_case_1_candidates(df_with_differences)
case_1_candidates
```

Thoughts: 

1.) Cases 1 - 3 will come more into play when we will be loading more new sheets from the company as these cases are highly related scope to new sheets. Note that we didn't had an extensive analysis for these three cases, unlike cases 4 and 5 where we add parameters in the matching condition one-at-a-time to quantify the surviving groupings, because Cases 1 - 3 still has the presence of FCC ASR and FAA registry number. BUT, that doesn't mean that we will not be exploring more for these three cases, we will after this.

2.) Case 4 is promising. But, it will not totally diminish the number of groupings in the prox audits. What most likely it will do is just do an automerge for each grouping that conforms to case 4's logic. That doesn't mean that the whole group will be away, because there will be instances where there are more than 2 records in a group. Say we have 5 records -- if one of the paired records and the reference record will be automerging, there will still be 3 more remaining records left untouched. Thus, the new group will now have 4 records. But at least this will be an easy "confirm correct and continue" by the analyst, most of the time. 

3.) Case 5 will come into play for cases where the data reported by the company is "not so usable" since there are no government identifier being reported for these tower assets (i.e., FCC ASR and FAA). Probably, these are tower assets that are not registered to any government entity.

4.) Also, we will explore more the outputs we have for cases 4 and 5 to design matching criteria that will give us more accurate results.
